## NumPy array and basics

Learn about numpy

```python
import numpy as np
```

### Creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("1D array :",arr_1d)
```

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D array :",arr_2d)
```

### List vs NumPy array

```python
py_list = [1,2,3]
print("Python list multiplication", py_list*2)
```

```python
arr = np.array([1,2,3])
print("Python array multiplication",arr*2)
```

```python
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("List Operation time:", time.time()-start)

start = time.time()
np_arr = np.arange(1000000)*2
print("NumPy Operation time:",time.time()-start)
```

### Creating array from scratch


```python
zero_arr = np.zeros((3,4))
print("zeros array : \n",zero_arr)
```

```python
ones_arr = np.ones((2,3))
print("Ones array: \n", ones_arr)
```

```python
full_arr = np.full((2,2),7)
print("Full array:\n",full_arr)
```

```python
random_arr = np.random.random((2,3))
print("Random array: \n ",random_arr)
```

```python
sequence_arr = np.arange(0,11,2)
print("Sequence array:\n",sequence_arr)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)
```

```python
matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix:",matrix)
```

```python
tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("Tensor:",tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
               [4,5,6]])

print("Shape:",arr.shape)
print("Dimension:",arr.ndim)
print("Size:",arr.size)
print("DType:",arr.dtype)
```

### Array reshaping

```python
arr = np.arange(12)
print("Original array :",arr)

reshaped = arr.reshape((3,4))
print("\nReshaped array :",reshaped)

flattened = reshaped.flatten()
print("\nFlattened array :",flattened)

# ravel (returns view, instead of copy)
raveled = reshaped.ravel()
print("\nRaveled array :",raveled)

# Transpose
transpose = reshaped.T
print("\nTranspose array :",transpose)
```
