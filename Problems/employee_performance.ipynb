### Employee Performance System

An HR system stores employee performance ratings (scale 1–5) for 100 employees over 4 quarters in a NumPy array.
Task:
- Normalize the ratings using min-max normalization
- Calculate the average rating per employee
- Identify employees whose average rating is above the company mean


```python
import numpy as np
```

```python
rng = np.random.default_rng(seed = 2)
employee_data = rng.integers(1,6,size = (4,100))
print("Ratings of Employee Quaterly:\n",employee_data)
```

```python
company_avg = employee_data.mean()
avg_rating = employee_data.mean(axis=0)
print("Average rating per employee:\n",avg_rating)
max_data = employee_data.max()
min_data = employee_data.min()
# Normalization used to range data between 0 to 1 with formula (data-min)/(max-min)
normalized_data = (employee_data - min_data)/(max_data - min_data)
print("Normalized data :",normalized_data)
```

```python
min_rate_idx = np.argmin(avg_rating)
max_rate_idx = np.argmax(avg_rating)
print("Index of Employee with minimum rating:",min_rate_idx)
print("Index of Employee with maximum rating:",max_rate_idx)
print("Employees whose average rating is above the company mean",avg_rating[avg_rating>company_avg])
```
