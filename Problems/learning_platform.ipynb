### Online Learning Platform
An online platform tracks time spent (in minutes) by students on 6 courses over 14 days using a NumPy array.
Task:
- Compute total time spent per course
- Identify courses where average daily engagement exceeds a threshold
- Rank courses based on engagement


```python
import numpy as np
```

```python
rng = np.random.default_rng(seed = 42)
course_time_data = rng.integers(0,121,(6,14))
print("Time spent on courses over 14 days:\n",course_time_data)
```

```python
time_per_course = course_time_data.sum(axis=1)
print("Total time spent per course :",time_per_course)
```

```python
avg_daily_engagement = np.sum(course_time_data,axis=1)
# let's take mean as threshold
threshold = avg_daily_engagement.mean()
mask = avg_daily_engagement>threshold
print("Courses average engagement where they exceeds a threshold:",avg_daily_engagement[mask])
```

```python
courses_ranking = time_per_course.argsort()[::-1]
print("Ranking of courses from 0 to 5:",courses_ranking)
```
