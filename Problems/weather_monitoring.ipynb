### Weather Monitoring System
A weather station stores hourly temperature readings for 7 days in a NumPy array of shape (7, 24).
Task:
- Find the daily maximum and minimum temperature
- Identify the day with the largest temperature variation
- Replace outliers beyond ±2 standard deviations with the mean temperature


```python
import numpy as np
```

```python
rng = np.random.default_rng(seed = 42)
temp_data = rng.integers(-30,40,size = (7,24))
print("Temperature data of 7 days for every hour:\n",temp_data)
```

```python
rows = np.arange(temp_data.shape[0])
daily_maximum = temp_data.argmax(axis = 1)
daily_minimum = temp_data.argmin(axis = 1)
print("Minimum temperature of every day:",temp_data[rows,daily_minimum])
print("Maximum temperature of every day:",temp_data[rows,daily_maximum])
```

```python
temperature_variation = temp_data[rows,daily_maximum] - temp_data[rows,daily_minimum]
print("Day with the largest temperature variation is ",temperature_variation.argmax()+1)
```

```python
# to change the outliers
data_mean = temp_data.mean()
data_deviation = temp_data.std()
lower_bound = data_mean - (2*data_deviation)
upper_bound = data_mean + (2*data_deviation)
bool_mask = (temp_data<lower_bound) | (temp_data>upper_bound)
temp_data[bool_mask] = data_mean
print("After removing outliers:\n",temp_data)
```
