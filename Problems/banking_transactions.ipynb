### Banking Transactions

A bank stores transaction amounts for 50 accounts across 20 days in a NumPy array.
Task:
- Separate debit and credit transactions
- Compute daily net balance changes
- Identify accounts with continuous negative balance trends

```python
import numpy as np
```

```python
rng = np.random.default_rng(seed = 2)
data = rng.integers(-10000,10000,size=(50,20))
data
```

### Debit Transaction

```python
debit = data[data<0]
debit
```

### Credit Transaction

```python
credit = data[data>0]
credit
```

### Daily net balance changes

```python
net_balance = np.diff(data,axis = 1)
net_balance
```

### Accounts with continious Negative balance trends

```python
negative_trend = np.all(net_balance,axis = 1)
account_indices = np.where(negative_trend)[0]
account_indices
```
