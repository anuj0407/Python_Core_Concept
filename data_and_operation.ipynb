## Advance operation with business examples


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("------ Zomato sales analysis ------")
print("\n Sales data shape:",sales_data.shape)
print("\n sample data for first 3 restraurant: \n",sales_data[:3])
```

```python
# total sales per year
print(np.sum(sales_data,axis=0))
yearly_total = np.sum(sales_data[:, 1:],axis = 0)
print("Yearly total :",yearly_total)

# Minimum sales per restraurant
min_sales = np.min(sales_data[:, 1:],axis = 1)
print("Minimum sales :",min_sales)

# Maximum sales per year 
max_sale = np.max(sales_data[:, 1:], axis = 0)
print("Maximum sales per year:",max_sale)

# Average sales per restraurant
avg_sales = np.mean(sales_data[:, 1:],axis = 1)
print("Average sales per year:",avg_sales)
```

```python
cumulative_sale = np.cumsum(sales_data[:, 1:], axis = 1)
print("Cumulative sales:\n",cumulative_sale)

plt.figure(figsize = (8,6))
plt.plot(np.mean(cumulative_sale,axis=0))
plt.title("Average Cumulative sales across all restraurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("Vector addition :", vector1+vector2)

print("\n Multiplication vaector :",vector1 * vector2 )

print("\n Dot Producct :",np.dot(vector1,vector2))

angle = np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("\n Angle :",angle)
```
